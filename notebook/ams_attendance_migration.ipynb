```python
import pandas as pd

# Load your CSV
df = pd.read_csv("../data/attendance/data-1775542390660.csv")

# Convert columns to datetime
cols = ["date", "checkIn", "createdAt"]
for col in cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')

# Set new base date
new_date = pd.Timestamp("2026-04-07")

# Function to update datetime
def update_datetime(dt):
    if pd.isna(dt):
        return dt
    new_dt = new_date + pd.Timedelta(
        hours=dt.hour,
        minutes=dt.minute,
        seconds=dt.second
    )
    return new_dt + pd.Timedelta(minutes=5)

# Apply transformation
for col in cols:
    df[col] = df[col].apply(update_datetime)

# ✅ Set default fields to NULL
null_columns = ["checkOut", "otHours", "deviceId", "shiftId", "totalHours"]

for col in null_columns:
    if col in df.columns:
        df[col] = "NULL"

# Save to new CSV
df.to_csv("../data/attendance/updated_data.csv", index=False)

print("✅ New CSV created: updated_data.csv")
```

```python
# import pandas as pd

# Load your updated CSV
df = pd.read_csv("../data/attendance/updated_data.csv")

# Replace NaN with None
df = df.where(pd.notnull(df), None)

# Function to format values for SQL
def format_value(val):
    if val is None:
        return "NULL"
    elif isinstance(val, str):
        return f"'{val}'"
    else:
        return f"'{val}'"  # for datetime also

# Build VALUES part
values_list = []

for _, row in df.iterrows():
    values = [
        format_value(row["userId"]),
        format_value(row["date"]),
        format_value(row["checkIn"]),
        format_value(row["checkOut"]),
        format_value(row["status"]),
        format_value(row["createdAt"]),
        format_value(row["otHours"]),
        format_value(row["deviceId"]),
        format_value(row["shiftId"]),
        format_value(row["totalHours"]),
    ]
    
    values_list.append(f"({', '.join(values)})")

# Combine into one query
query = f"""
INSERT INTO "Attendance" (
    "userId", date, "checkIn", "checkOut", status, "createdAt",
    "otHours", "deviceId", "shiftId", "totalHours"
)
VALUES
{',\n'.join(values_list)};
"""

# Save to file (optional)
with open("insert_attendance.sql", "w") as f:
    f.write(query)

print("✅ SQL file generated: insert_attendance.sql")

# Print preview
print(query[:1000])  # first 1000 chars
```

```python
import pandas as pd
import math

# Load CSV
df = pd.read_csv("../data/attendance/updated_data.csv")

# Function to format SQL values
def format_value(val):
    # Handle NaN / None
    if val is None or (isinstance(val, float) and math.isnan(val)):
        return "NULL"
    
    # Handle strings
    if isinstance(val, str):
        return f"'{val}'"
    
    # Handle datetime (already string from CSV)
    return f"'{val}'"

# Build VALUES
values_list = []

for _, row in df.iterrows():
    values = [
        format_value(row["userId"]),
        format_value(row["date"]),
        format_value(row["checkIn"]),
        format_value(row["checkOut"]),
        format_value(row["status"]),
        format_value(row["createdAt"]),
        format_value(row["otHours"]),
        format_value(row["deviceId"]),
        format_value(row["shiftId"]),
        format_value(row["totalHours"]),
    ]
    
    values_list.append(f"({', '.join(values)})")

# Final query
query = f"""
INSERT INTO "Attendance" (
    "userId", date, "checkIn", "checkOut", status, "createdAt",
    "otHours", "deviceId", "shiftId", "totalHours"
)
VALUES
{',\n'.join(values_list)};
"""

# Save file
with open("insert_attendance.sql", "w") as f:
    f.write(query)

print("✅ Fixed SQL generated (no 'nan' issue)")
```
